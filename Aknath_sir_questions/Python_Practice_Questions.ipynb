**Problem Statement:**

divisible by 4 but not divisible by 100.

**Input:**
A single integer representing the year.

**Output:**
Print Leap Year or Not a Leap Year.

**Examples:**
```text
Input:
Year = 2000
Output:
Leap Year
Explanation:
2000 is divisible by 400.
Input:
Year = 1900
Output:
Not a Leap Year
Explanation:
1900 is divisible by 100 but not by 400.
```


````python
# Solution for Question 5: Check Whether a Year is a Leap Year
# Write your code below:
num = int(input("Enter the Year "))
if num % 4 == 0:
    if num % 100 == 0:
        if num % 400 == 0:
            print("Leap year")
        else:
            print("Not a leap year")
    else:
        print("leap year")
else:
    print("Not a leap year")
````
