#  Difference between Matplotlib and Seaborn 

# matplot : 
1. It is used to plot simple data.
2. It has limited Plots.
3. It has attributes.

# Seaborn 
1. It is used to plot complex data.
2. It has more number of Plots.
3. It has more number of attributes.

# Types of Plots in Seaborn .

    1. Distribution Plots
    2. Chatogorical Plots 
    3. Matrix Plots 
    4. Regression Plots

# Types of Plots based on number of columns used in plot.

    1. Uni-variate Plot : It allows us to plot the graph based on one column.
    2. Bi-variate Plot :  It allows us to plot the graph based on two column.
    3. Multi-variate Plot : It allows us to plot the graph based on multiple column.

# 1. Distributed Plot : 

     It helps us to understand how the data got distributed   

----> Distributed -- Spread of values.    
     By the what can we analyse ?  
     --> min, max, highly dense value , unique values, venv/#

```python
import seaborn as sns 
import pandas as pd
import matplotlib as plt
import matplotlib.pyplot as plt
```

```python
ti=pd.read_csv(r"/Users/kunalsharma/Desktop/Ab-hoge-Python/Data-Analysis-Datasets/Datasets-Practice/tips.csv")
```

```python
sns.histplot(x="sex",data=ti,color="orange")
```

```python
# palette-  inferno,viridis, magma, Set1,Set2,Set3,coolwarm, Reds, Green, Purples
a=sns.histplot(x="sex",data=ti,palette="Set1",hue="smoker")
```

```python
# We can use loop for showing the bar levels
a=sns.histplot(x="sex",data=ti,palette="viridis",hue="smoker")
for i in a.containers:
    a.bar_label(i)
```

```python


qu = pd.read_csv(
    r"/Users/kunalsharma/Desktop/Ab-hoge-Python/Data-Analysis-Datasets/Datasets-Practice/winequalityN.csv"
)

sns.histplot(
    data=qu,
    x="quality",
    hue="type",
    palette="Reds",
    discrete=True,
    multiple="dodge",
    shrink=0.8,
)

plt.show()
```

```python
hr=pd.read_csv(r"/Users/kunalsharma/Desktop/Ab-hoge-Python/Data-Analysis-Datasets/Datasets-Practice/hr_data.csv")
a=sns.histplot(x="salary",data=hr,hue="salary")
for i in a.containers:
    a.bar_label(i,label_type="center",size=15)
```
