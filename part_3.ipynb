```python
# "Write a Python program to find the missing number from a list containing ‭$1$‬ to ‭$N$‬ consecutive integers."
num = [1,2,3,4,6,7,8,9,10]
n = len(num)+1
e_total = n*(n+1)/2
print(e_total)
c_total = sum(num)
print(c_total)
f_answer = (e_total-c_total)
print(f_answer)
```

```python
tokens = [3,32,32,23,23,23,23,2323,32,43,4,5,24,24,4,54]

n = len(tokens)+1
all_total = n*(n+1)/2
print("This is the total sum of all tokens",all_total)

c_total = sum(tokens)
print("This is the current total was given by students",c_total)

M_tokens = (all_total-c_total)
print("This is the missing Tokens",M_tokens)
```

```python
# Convert input string to an integer
length = int(input("Enter the number of terms: "))

n1, n2 = 0, 1

if length <= 0:
    print("Please enter a positive integer.")
elif length == 1:
    print(n1)
else:
    print(n1, n2, end=" ")
    for i in range(length - 2):
        next_num = n1 + n2
        print(next_num, end=" ")
        n1 = n2
        n2 = next_num
```
