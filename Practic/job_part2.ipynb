# wap to solve the given below requriment.

# O/P = aaaabbbbcc

```python
s = input("Enter the string")
output=" "
for i in s:
    if i.isalpha():
        x=i
else:
    d=int(i)
    output=output+(x*d)
res = sorted(output)
res=" ".join
print(res)

```

```python
# WAP to count the occurrence of each character in the given string
s = input("Enter the string")
char_count = {}
for char in s:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1
for char, count in char_count.items():
    print(f"{char}: {count}")
```

```python
# WAP TO SOLVE THE GIVEN BELOW REQURMENT
# IP = 3A2B1C
# OP = 3A2B1C
s = input("Enter the string")
d={}
output=" "
for i in s:
    d[i]=d.get(i,0)+1 # imp hai broo
for i in d:
    output=output+str(d[i])+i
print(output)
```

```python
# WAP TO TAKE A STRING AS INPUT ASSUMED INPUT STRING TO CONTAINS ONLY ALPHABET OR DIGITM, WAP A CODE TO SHORT CHAR OF THE STRING.FIRST APLHABET SYMBOL FOLLOWED BY DIGIT.
# IP = D5A2B1
#OP = ABD125
```

```python
S = input("enter the kuch bhee")
print(S)

alpha=[]
digit = []
for i in S:
     if i.isalpha:
          alpha.append[i]
else:
     digit.append[i]
output ="".join(sorted(alpha)+sorted(digit))
print(output)
```
