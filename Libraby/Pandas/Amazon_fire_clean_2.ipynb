```python
import pandas as pd
df = pd.read_csv('Data-Analysis-Datasets/Datasets-Practice/amazon_fires.csv',encoding = 'ISO-8859-1', sep=',', na_values=[''], quotechar='"')
amazon_fires
```

```python
print(df.head())
print(df.isnull().sum())
print(df.info())
```

```python
df = df.rename(columns={
    'ano': 'Year',
    'mes': 'Month',
    'estado': 'State',
    'numero': 'Number_of_Fires',
    'encontro': 'Date'
})
```

```python
# ' Fires' word ko blank '' se replace kara hai
df['Number_of_Fires'] = df['Number_of_Fires'].astype(str).str.replace(' Fires', '')

# Ab isko number me convert kar do (agar kuch aur kachra hoga to 'coerce' usko NaN bana dega)
df['Number_of_Fires'] = pd.to_numeric(df['Number_of_Fires'], errors='coerce')
```

```python
# Hame pata hai 132 jagah data missing hai, unhe hum 0 (zero) se bhar denge
df['Number_of_Fires'] = df['Number_of_Fires'].fillna(0)
```

```python
# meri matlab data ke corrected dictionary
amz = {
    'Janeiro': 'Jan', 'Fevereiro': 'Feb', 'Março': 'Mar',
    'Abril': 'Apr', 'Maio': 'May', 'Junho': 'Jun',
    'Julho': 'Jul', 'Agosto': 'Aug', 'Setembro': 'Sep',
    'Outubro': 'Oct', 'Novembro': 'Nov', 'Dezembro': 'Dec'
}

# .map() use karke saare naam replace kar do
df['Month'] = df['Month'].map(amz)
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```
