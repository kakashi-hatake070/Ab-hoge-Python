## 1. Import Libraries

NumPy is used for numerical arrays and calculations. Pandas is used for Series, DataFrames, cleaning, and analysis.

```python
import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

## 2. NumPy Arrays

A NumPy array stores values in a structured numerical object. Arrays can be one-dimensional, two-dimensional, or multi-dimensional.

**Important:** NumPy converts mixed values to a compatible common data type.

```python
# Create arrays from a list, a list literal, and a tuple.
values = [5, 10, 15, 20]
arr = np.array(values)
arr2 = np.array([5, 1, 22, 44])
mixed_array = np.array((True, False, 3, 2.5, 2 - 3j, "hi"))

print("Array from list:", arr)
print("Array from literal:", arr2)
print("Mixed array:", mixed_array)

# Convert multiple Python lists into NumPy arrays.
first_list = [1, 2, 3, 4, 5]
second_list = [0, 9, 8, 7, 6]
first_array = np.array(first_list)
second_array = np.array(second_list)
first_array, second_array
```

## 3. Dimensions and Reshaping

- `ndim` returns the number of dimensions.
- `shape` returns the size of each dimension.
- `reshape(rows, columns)` changes the shape without changing the data.
- `-1` lets NumPy calculate one dimension automatically.

```python
one_dimensional = np.array([1, 2, 3, 4, 5])
two_dimensional = np.arange(1, 16).reshape(3, 5)
three_dimensional = np.arange(1, 19).reshape(2, 3, 3)

print("1D array:\n", one_dimensional)
print("2D array:\n", two_dimensional)
print("3D array:\n", three_dimensional)

numbers = np.arange(1, 25)
rows_and_columns = numbers.reshape(4, 6)
blocks = numbers.reshape(2, 2, 6)

print("Dimensions:", numbers.ndim, rows_and_columns.ndim, blocks.ndim)
print("Shapes:", numbers.shape, rows_and_columns.shape, blocks.shape)

print("Values from 1 to 12:", np.arange(1, 13))
print("Evenly spaced values:", np.linspace(1, 10, 5))
```

## 4. Array Creation and Filtering

`zeros`, `identity`, and `eye` create useful matrix patterns. `where` applies a condition element by element, while Boolean filtering returns only values that match a condition.

```python
# Create arrays filled with zeros and identity-style matrices.
print("Zeros:\n", np.zeros((3, 4)))
print("Identity:\n", np.identity(4))
print("Eye matrix:\n", np.eye(3, 5))

# Random arrays.
print("Random integers:", np.random.randint(1, 3, 5))
print("Random values between 0 and 1:", np.random.rand(4))
print("Random normal values:", np.random.randn(4))

numbers = np.arange(1, 13).reshape(4, 3)
print("Even or odd:\n", np.where(numbers % 2 == 0, "even", "odd"))
print("Multiples of 5 replaced with 0:\n", np.where(numbers % 5 == 0, 0, numbers))

ages = np.array([[35, 44, 23, 67, 87, 45, 23, 45, 23],
                 [87, 56, 84, 34, 25, 54, 34, 65, 54]])
print("Ages above 60:", ages[ages > 60])
print("Ages between 25 and 50:", ages[(ages > 25) & (ages < 50)])
```

## 5. Pandas Series

A Series is a one-dimensional labelled structure.

- Use `.loc` for label-based selection.
- Use `.iloc` for position-based selection.
- A Series can be created from a list or dictionary.

```python
scores = pd.Series(
    [5, 10, 15, 20, 25],
    index=["a", "b", "c", "d", "e"],
    name="Scores",
)
print(scores)

fruit_prices = pd.Series(
    [100, 50, 400, 800],
    index=["apple", "banana", "kela", "aam"],
    name="Price",
)
print(fruit_prices)

employee_counts = pd.Series({
    "Apple": 100,
    "Banana": 50,
    "Kela": 400,
    "Aam": 800,
})

print("Using labels:\n", employee_counts.loc[["Apple", "Banana", "Kela"]])
print("Using positions:\n", employee_counts.iloc[[2, 3, 1]])
```

```python
# In Pandas, the primary difference between .loc and .iloc is how they select data. The .loc method uses label-based indexing to select rows and columns by their actual names, while .iloc uses integer-based indexing to select data by its numerical position.
# Here is a comparison of their key differences:
# FEATURE	.LOC	.ILOC
# Selection Type	Label-based	Integer-based
# Input Values	Row and column labels	Numerical indices starting from 0
# Slicing End	Inclusive of the end label	Exclusive of the end index
# Error Handling	Raises a KeyError	Raises an IndexError
# Slicing Behavior
# When slicing a DataFrame, .loc includes the stop label in the selection, whereas .iloc excludes the stop index. This difference is a common source of off-by-one selection mistakes.
# Error Handling
# If you try to select a non-existent label with .loc, it raises a KeyError. Conversely, if you try to select an out-of-bounds numerical position with .iloc, it raises an IndexError.
# Boolean Indexing
# The .loc method supports boolean indexing directly, allowing you to filter data based on conditions. The .iloc method does not support boolean indexing in the same flexible way and expects a boolean array-like object representing row positions.
```
