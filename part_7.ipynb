```python
# Wap to calculate lcm for a give two numbers
```

```python
n1 = int(input("Enter a number: "))
n2 = int(input("Enter another number: "))
if n1<n2:
    small=n1
else:
    small=n2
for i in range (1,small+1):
    if n1%i==0 and n2%i==0:
        hcf=i
lcm=(n1*n2)/hcf
print(lcm)
```

```python
# wap to print to check whether a given numbers are co-prime or not.
```

```python
num1 = int(input("Enter a number: "))
num2 = int(input("Enter a second number: "))

if num1 < num2:
    small = num1
else:
    small = num2

hcf = 1  # Initialize hcf

for i in range(1, small + 1):
    if num1 % i == 0 and num2 % i == 0:  # Fixed: changed 2 to i
        hcf = i

if hcf == 1:
    print("The numbers are co-prime")
else:
    print("The numbers are not co-prime")
```

```python
# CO-primr [  the numbers are said to be co-prime to each other if there hcf is equal to 1 ]
```

```python
# wap to check whether number is perfect number or not.
```

```python
num = int(input("Enter a number: "))
sum = 0
for i in range(1,num+1):
    if num % i == 0:
        sum=sum+i
if num == sum:
    print("The number is co prime")
else:
    print("The number is not co prime")
```

```python
# wap to remove duplicate element or repeated elements from the given list
```

```python
n = eval(input("Enter a number: "))
res=[]
for i in n:
    if i not in res:
        res.append(i)
print(res)
```
