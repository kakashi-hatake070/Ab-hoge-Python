```python
n = int(input("Enter the rows"))
k = 1
for i in range (n):
    for j in range(n):
        print(f"{k:3}",end=" ")     # f"{k:2}" this is use for formate the output.
        k = k+1
    print()
```

```python
n = int(input("Enter the rows"))
k = n*n
for i in range (n):
    for j in range(n):
        print(f"{k:3}",end=" ")     # f"{k:2}" this is use for formate the output.
        k = k-1
    print()
```
